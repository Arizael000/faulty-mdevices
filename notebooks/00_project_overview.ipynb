# Faulty Medical Devices ? Project overview


## Dataset

Large collection of recalls, safety alerts and field safety notices about medical devices distributed worldwide. Published by ICIJ and available on Kaggle as the *Faulty Medical Devices Global Dataset*.

| File | Raw shape | Clean shape | Role |
|---|---:|---:|---|
| `devices` | 118,249 ? 15 | 30,556 ? 12 | Device master data |
| `events` | 124,969 ? 30 | 39,520 ? 15 | Recall events |
| `manufacturers` | 31,827 ? 10 | 31,790 ? 6 | Manufacturer dimension |
| events + devices | ? | 32,224 matched | Device-linked recalls |


## Recall classification (FDA)

- **Class I**: a reasonable chance that the product will cause serious health problems or death.
- **Class II**: the product may cause a temporary or reversible health problem, or a slight chance of serious harm or death.
- **Class III**: the product is not likely to cause any health problem or injury.

**HDE** (*Humanitarian Device Exemption*) is not a risk class: it is a special FDA approval pathway for high-risk devices used in small patient populations, so it is reported next to Classes I?III.


## Business questions

1. Which manufacturers accumulate the most faulty devices, and what kind of devices are they?
2. Which devices appear most frequently in the dataset?
3. Which risk classification is the most prevalent?
4. Which country or distribution zone is the most prevalent?
5. When did the recalls happen, and what causes drive them?


## Final architecture

```
data/raw/            original ICIJ CSVs
data/processed/      cleaned tables + eventsEnriched.csv + etl_report.json
src/                 reproducible ETL pipeline (extract ? transform ? merge ? metrics)
dashboard/           index.html + styles.css + app.js (static, data embedded)
notebooks/           exploration and cleaning documentation
```

Run `python -m src.pipeline` to rebuild every processed artefact and `dashboard/app.js`. Then open `dashboard/index.html`. See the repository `README.md` for details.


## Key results

- 39,520 recalls after cleaning, spanning 1997?2018.
- 32,224 recalls (81.5%) link to a cleaned device record; these power the risk, manufacturer, category, device and implant views.
- Class II is the dominant risk class, with 24,256 device-linked recalls; Class I has 5,589 and Class III 2,094.
- 7,238 device-linked recalls involve an implanted device.
- Customed, Inc. is the manufacturer with the most recalls (695), followed by Siemens Medical Solutions USA (481) and Stryker Howmedica Osteonics (476).
- Orthopedic Devices is the most recalled product category (5,397 recalls), followed by Cardiovascular Devices (4,203).
- 2017 is the peak year (2,773 recalls).
- The full events table covers four countries (United States, Australia, United Kingdom, India); the device-linked subset is United States only because complete device rows come from the FDA source.


## Limitations

- The clean device table comes from a complete-case filter, so the enriched subset is biased toward US FDA records with fully populated fields.
- 973 recalls have no usable `date_initiated_by_firm`; they appear in totals but not in the year trend.
- Manufacturer names are not unique identifiers; counts are computed by `manufacturer_id`.
- Free-text causes and device names required normalization (top 15 values plus a grouped tail).
