About the data set  
This is a large collection of recalls, safety alerts and field safety notices about medical devices distributed worldwide. The dataset has been published by ICIJ.

Classification of recalls (from FDA website):

Class I: A situation where there is a reasonable chance that a product will cause serious health problems or death.

Class II: A situation where a product may cause a temporary or reversible health problem or where there is a slight chance that it will cause serious health problems or death.

Class III: A situation where a product is not likely to cause any health problem or injury.

La clasificación HDE son de carácter humanitario, son recall de dispositivos con este
tipo de aprobación para una cantidad reducida de pacientes, estos dispositivos no
se someten a pruebas masivas.

HDE significa Humanitarian Device Exemption (Exención de Dispositivos Humanitarios).

No es una "clase de riesgo" estándar como la Clase I, II o III, sino una vía regulatoria especial (o una clasificación de aprobación) para un grupo muy específico de productos médicos de alto riesgo (generalmente catalogados como Clase III). En Estados Unidos esto lo aprueba la FDA.

The objective of this to build an inform in PowerBi related to the information in the dataset.

I hope to solve this initial questions:
- From wich manufacturer is the devices that fails the most and what kind of device is.
- Wich devices appears more frequent in the dataset.
- Wich risk classification is the most prevalent in the dataset.
- Wich country or zone ditributed to is the most prevalent in the dataset.

Note that I will set a limit of 30% of nulls quantity in columns. 

Irrelevant columns will be droped for this analysis, such as links to Companys o Goverment.

In this kind of analysis, imputation of data is not possible because the majority of the data consists of text, links, and coded names. Therefore, the main flow of the project will involve cleaning the data to upload it to a database. Then, Power BI will be used to create a dashboard that helps extract valuable and logical insights from the database.
